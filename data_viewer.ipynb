```python
import h5py
import pandas as pd
import numpy as np

data_path = 'CHAP2/example_data_wrist/2018-06-25.h5'

with h5py.File(data_path, 'r') as f:
    # One row per window, with summary stats or raw axes
    labels = f['label'][:]
    non_wear = f['non_wear'][:]
    sleeping = f['sleeping'][:]
    time = f['time'][:]
    data = f['data'][:]  # (358, 100, 3)

# Option 1: Per-window summary (mean of each axis)
df = pd.DataFrame({
    'time': time,
    'label': labels,
    'non_wear': non_wear,
    'sleeping': sleeping,
    'x_mean': data[:, :, 0].mean(axis=1),
    'y_mean': data[:, :, 1].mean(axis=1),
    'z_mean': data[:, :, 2].mean(axis=1),
})
print(df.head())
```
